## 1. Importing Library

```python
import pandas as pd
```

```python
# loading dataset 
df = pd.read_csv('pandas_dataset.csv')

# Initial Data Shape
shape_before = df.shape
print("Before Data Shape:")
print(shape_before)
print("\n")

# Displaying the first 5 rows
print("First Five Rows:")
print(df.head())
```

## 2. Renaming Columns

```python
# rename columns to clean format: remove spaces, convert to lowercase
df.columns = df.columns.str.replace(' ', '_').str.lower()

print("Cleaned Column Names:")
print(df.columns.tolist())
```

## 3. Dropping Unnecessary Rows

```python
# drop the extra rows above the header
df = df.iloc[1:]

# reset the index after cleaning
df = df.reset_index(drop=True)
print("First 5 rows (Cleaned):")
print(df.head())
```

## 4. Dropping Unecessary Columns

```python
# order_date is not used anywhere in the analysis as no date filtering or grouping needed, so it is safe to drop

df = df.drop('order_date', axis=1)
print(df.columns)
```

## 5. Counting Missing Values

```python
print("No. of Null values per column:")
print(df.isnull().sum())
```

## 6. Handling Missing Values

**a. Drop Rows Based on a Column**

```python
df = df.dropna(subset=['sales'])
print(df.shape)
```

**b. Fill Missing Values**

```python
# Fill missing values in profit using mean, discount using median

df['profit'] = df['profit'].fillna(df['profit'].mean())
df['discount'] = df['discount'].fillna(df['discount'].median())
print(df[['profit', 'discount']].isnull().sum())
```

## 7. Handling Missing Categorical Data

```python
# Replace missing customer_name values with 'Unknown'

df['customer_name'] = df['customer_name'].fillna('Unknown')
print(df['customer_name'].isnull().sum())
```

## 8. Detect and Remove Duplicates

```python
# identify duplicate rows based on order_id using df.duplicated

print(df.duplicated(subset=['order_id']).sum())
```

```python
# remove duplicates and show how many were removed

before = df.shape[0]
df = df.drop_duplicates(subset=['order_id'])
print(before - df.shape[0])
```

## 9. Filtering and creating new columns

```python
# Display all orders where unit_price is greater than 20,000

print(df[df['unit_price'] > 20000])
```

```python
# Display customer_name, category, status where unit_price > 10,000 and status == 'Completed'

print(df[(df['unit_price'] > 10000) & (df['status'] == 'Completed')][['customer_name', 'category', 'status']])
```

```python
# Create a new column total_amount = quantity x unit_price - discount

df['total_amount'] = df['quantity'] * df['unit_price'] - df['discount']
print(df.head())
```

```python
# Create a column customer_type: Bulk Buyer if quantity >= 3, else Regular Buyer

df['customer_type'] = df['quantity'].apply(lambda x: 'Bulk Buyer' if x >= 3 else 'Regular Buyer')
print(df.head())
```

## 9. Final Clean Dataset

```python
# Shape of dataset before and after cleaning

print(shape_before)
print(df.shape)
```
