### 4.4 Invalid or inconsistent values

```python
# does the result column match the score?
wrong_h = results[(results["home_goals"] > results["away_goals"]) & (results["result"] != "H")]
wrong_a = results[(results["home_goals"] < results["away_goals"]) & (results["result"] != "A")]
wrong_d = results[(results["home_goals"] == results["away_goals"]) & (results["result"] != "D")]
print("wrong results:", len(wrong_h) + len(wrong_a) + len(wrong_d))

# negative goals, team playing itself
print("negative goals:", (results["home_goals"] < 0).sum())
print("same home and away team:", (results["home_team"] == results["away_team"]).sum())
```

```python
# matches per season (should be 380)
print(results.groupby("season").size())
```

```python
# team names in both files
print(set(results["home_team"]) == set(stats["team"]))
```

## 5. EDA - Exploring the data

### 5.1 Home wins, draws and away wins

```python
print(results["result"].value_counts())
print()
print(results["result"].value_counts(normalize=True) * 100)
```

```python
results["result"].value_counts().plot(kind="bar")
plt.title("Match results (H = home win, A = away win, D = draw)")
plt.show()
```

### 5.2 Average goals per match in each season

```python
avg_goals = results.groupby("season")["total_goals"].mean()
avg_goals.plot(kind="line", marker="o")
plt.title("Average goals per match")
plt.xticks(rotation=45)
plt.show()
```

### 5.3 Teams with the most wins

```python
top_wins = stats.groupby("team")["wins"].sum().sort_values(ascending=False).head(10)
top_wins.plot(kind="bar")
plt.title("Top 10 teams by total wins (12 seasons)")
plt.show()
```

### 5.4 Goals scored and conceded

```python
stats["goal_difference"] = stats["goals"] - stats["goals_conceded"]
stats.sort_values("goal_difference", ascending=False)[["team", "season", "goals", "goals_conceded", "goal_difference"]].head(5)
```

```python
team_goals = stats.groupby("team").agg({"goals": "sum", "goals_conceded": "sum", "wins": "sum"})
team_goals.sort_values("goals", ascending=False).head(5)
```

### 5.6 Do more shots on target mean more wins?

```python
plt.scatter(stats["ontarget_scoring_att"], stats["wins"])
plt.xlabel("Shots on target")
plt.ylabel("Wins")
plt.show()

print(stats["ontarget_scoring_att"].corr(stats["wins"]))
```
